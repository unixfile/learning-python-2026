## if/elif/else

```python
password = "12345"

if password == "12345":
    print("That's a bad password!")
```

```python
temperature = 8

if temperature > 20:
    print("It's warm")
elif temperature > 10:
    print("It's cool")
elif temperature > 5:
    print("It's cold")
else:
    print("It's normal Swedish temperature")
```
