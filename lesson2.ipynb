## Files

```python
f = open("mockup.txt")
content = f.read()
f.close()
type(content), len(content)
```

```python
f = open("mockup.txt")
lines = f.readlines()
f.close()
type(lines), len(lines)
```

```python
f = open("mockup.txt")
for line in f:
    print(line, end="")
f.close()
```

```python
f = open("hello.txt", "w")
f.write("Travel far, ")
f.write("meet yourself.\n")
f.flush()
f.close()
```

```python
with open("hello.txt", "r") as f:
    print(f.read())
```

```python
with open("hello.txt", "a") as f:
    f.write("Appending...\n")

with open("hello.txt", "r") as f:
    print(f.read())
```
