## Slicing

```python
s = [0, 1, 2, 3, 4, 5]
s[1:4]  # includes 1, excludes 4
```

```python
s[:3], s[3:]
```

```python
s[::2], s[::-1]
```

```python
"banan"[0:3], "banan"[-1]  # strings index and slice too
```

## for loop

```python
fruits = ["banan", "melon", "kiwi", "citron"]
for fruit in fruits:
    print(fruit)
```

## break and continue

```python
for n in range(1, 10):
    if n == 5:
        break
    print(n, end=' ')
```

```python
for n in range(1, 10):
    if n % 2 == 0:
        continue
    print(n, end=' ')
```

## enumerate

```python
fruits = ["banan", "melon", "kiwi", "citron"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

## Copying lists

```python
a = [1, 2, 3]
b = a
b.append(4)
a
```

```python
c = a.copy()
c.append(5)
a
```
