## Set idiosyncrasies

```python
fruits = {'banan', 'banan', 'melon', 'kiwi'}
fruits
```

## Adding and removing

```python
countries = {'Brasilien', 'Chile'}
countries.add('Bolivia')
countries
```

```python
countries.update({'Brasilien', 'Paraguay'})
countries
```

```python
try:
    countries.remove('Sverige')
except KeyError as e:
    print(f"KeyError: {e}")
countries.discard('Sverige')
```

## Set operations

```python
low = {1, 2, 3, 4}
high = {3, 4, 5, 6}
```

```python
low | high  # union
```

```python
low & high  # intersection
```

```python
low ^ high  # symmetric difference
```

```python
low - high  # difference
```

## Set comprehensions

```python
squares = {x**2 for x in range(6)}
squares
```

```python
evens = {x for x in range(10) if x % 2 == 0}
evens
```

## Dictionaries

```python
fruit_colors = {
    'äpple':    'grön',
    'banan':    'gul',
    'vindruva': 'lila',
}
type(fruit_colors)
```

## Accessing values

```python
fruit_colors['banan']
```

```python
try:
    fruit_colors['citron']
except KeyError as e:
    print(f"KeyError: {e}")
```

```python
fruit_colors.get('citron')
```

```python
fruit_colors.get('citron', 'okänd')
```

## Looping over dictionaries

```python
for fruit in fruit_colors:
    print(fruit)
```

```python
for color in fruit_colors.values():
    print(color)
```

```python
for fruit, color in fruit_colors.items():
    print(f"{fruit}: {color}")
```

## Modifying dictionaries

```python
fruit_colors['äpple'] = 'röd'
fruit_colors['mango'] = 'orange'
fruit_colors
```

```python
fruit_colors.pop('banan')
```

```python
fruit_colors.update({'kiwi': 'grön', 'äpple': 'grön'})
fruit_colors
```

## Nested dictionaries

```python
betyg = {
    'Alice': {'Matte': 85, 'Kemi': 92},
    'Bob':   {'Matte': 78, 'Kemi': 88},
}
betyg['Alice']['Matte']
```

```python
betyg['Bob']['Engelska'] = 87
betyg
```

## Dictionary comprehensions

```python
squares = {x: x**2 for x in range(5)}
squares
```

```python
original = {'äpple': 1, 'banan': 2, 'körsbär': 3}
swapped = {v: k for k, v in original.items()}
swapped
```

## Exception handling

```python
numbers = [1, 2, 3]

try:
    print(numbers[5])
except IndexError:
    print("Index out of range")
```

## Multiple exception types

```python
try:
    value = int("abc")
    numbers = [1, 2, 3]
    print(numbers[5])
except ValueError:
    print("Not a valid number")
except IndexError:
    print("Index out of range")
```

```python
try:
    value = int("abc")
except (ValueError, IndexError) as e:
    print(f"An error occurred: {e}")
```
