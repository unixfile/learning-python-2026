## Files

```python
f = open("mockup.txt")
content = f.read()
f.close()
type(content), len(content)
```

```python
f = open("mockup.txt")
lines = f.readlines()
f.close()
type(lines), len(lines)
```

```python
f = open("mockup.txt")
for line in f:
    print(line, end="")
f.close()
```

```python
f = open("hello.txt", "w")
f.write("Travel far, ")
f.write("meet yourself.\n")
f.flush()
f.close()
```

```python
with open("hello.txt", "r") as f:
    print(f.read())
```

```python
with open("hello.txt", "a") as f:
    f.write("Appending...\n")

with open("hello.txt", "r") as f:
    print(f.read())
```

## Adding and removing

```python
countries = {'Brasilien', 'Chile'}
countries.add('Bolivia')
countries
```

```python
countries.update({'Brasilien', 'Paraguay'})
countries
```

```python
try:
    countries.remove('Sverige')
except KeyError as e:
    print(f"KeyError: {e}")
countries.discard('Sverige')
```

## Dictionaries

```python
fruit_colors = {
    'äpple':    'grön',
    'banan':    'gul',
    'vindruva': 'lila',
}
type(fruit_colors)
```

## Accessing values

```python
fruit_colors['banan']
```

```python
try:
    fruit_colors['citron']
except KeyError as e:
    print(f"KeyError: {e}")
```

```python
fruit_colors.get('citron')
```

```python
fruit_colors.get('citron', 'okänd')
```

## Looping over dictionaries

```python
for fruit in fruit_colors:
    print(fruit)
```

```python
for color in fruit_colors.values():
    print(color)
```

```python
for fruit, color in fruit_colors.items():
    print(f"{fruit}: {color}")
```

## Modifying dictionaries

```python
fruit_colors['äpple'] = 'röd'
fruit_colors['mango'] = 'orange'
fruit_colors
```

```python
fruit_colors.pop('banan')
```

```python
fruit_colors.update({'kiwi': 'grön', 'äpple': 'grön'})
fruit_colors
```

## Nested dictionaries

```python
betyg = {
    'Alice': {'Matte': 85, 'Kemi': 92},
    'Bob':   {'Matte': 78, 'Kemi': 88},
}
betyg['Alice']['Matte']
```

```python
betyg['Bob']['Engelska'] = 87
betyg
```
