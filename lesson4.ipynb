## if/elif/else

```python
password = "12345"

if password == "12345":
    print("That's a bad password!")
```

```python
temperature = 8

if temperature > 20:
    print("It's warm")
elif temperature > 10:
    print("It's cool")
elif temperature > 5:
    print("It's cold")
else:
    print("It's normal Swedish temperature")
```

## for loop

```python
fruits = ["banan", "melon", "kiwi", "citron"]
for fruit in fruits:
    print(fruit)
```

## break and continue

```python
for n in range(1, 10):
    if n == 5:
        break
    print(n, end=' ')
```

```python
for n in range(1, 10):
    if n % 2 == 0:
        continue
    print(n, end=' ')
```

## enumerate

```python
fruits = ["banan", "melon", "kiwi", "citron"]
for index, fruit in enumerate(fruits):
    print(f"{index}: {fruit}")
```

## while loop

```python
i = 5
while i > 0:
    print(i, end=' ')
    i -= 1
```

## List comprehensions

```python
squares = [x**2 for x in range(6)]
squares
```

```python
evens = [x for x in range(10) if x % 2 == 0]
evens
```

## Generator expressions

```python
squares = (x**2 for x in range(100_000_000))
type(squares)
```

```python
next(squares), next(squares), next(squares)
```
